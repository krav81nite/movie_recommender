```python
import pandas as pd

ratings = pd.read_csv('data/processed/ratings_25m_clean.csv')

user_counts = ratings.groupby('userId').size()
active_users = user_counts[user_counts >= 50].index
ratings_small = ratings[ratings['userId'].isin(active_users)]

print(f"Reduced from {len(ratings):,} to {len(ratings_small):,} rows")
ratings_small.to_csv('data/processed/ratings_small.csv', index=False)
```

```python
user_counts = ratings.groupby('userId').size()
active_users = user_counts[user_counts >= 200].index
ratings_small = ratings[ratings['userId'].isin(active_users)]

print(f"Reduced from {len(ratings):,} to {len(ratings_small):,} rows")
ratings_small.to_csv('data/processed/ratings_small.csv', index=False)
```

```python
ratings.to_parquet('data/processed/ratings_25m_clean.parquet', index=False)
movies = pd.read_csv('data/processed/movies_enriched.csv')
movies.to_parquet('data/processed/movies_enriched.parquet', index=False)

import os
print(f"ratings parquet: {os.path.getsize('data/processed/ratings_25m_clean.parquet') / 1024 / 1024:.1f} MB")
print(f"movies parquet:  {os.path.getsize('data/processed/movies_enriched.parquet') / 1024 / 1024:.1f} MB")
print(f"ratings csv:     {os.path.getsize('data/processed/ratings_25m_clean.csv') / 1024 / 1024:.1f} MB")
```
